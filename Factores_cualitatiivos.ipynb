```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_excel('aguasfeo.xlsx')
data.describe
```

```python
#remplazando los valores de las variables cualitativas por valores numericos
tamano = data['Tamano'].replace('chica', 0).replace('grande', 1)
hielo = data['Hielo'].replace('no', 0).replace('si', 1)
```

```python
data[data['Sabor']=='Naranja']
```

```python
data['Naranja'] = np.where(data['Sabor'] == 'Naranja', 1, 0)
data['Coco'] = np.where(data['Sabor'] == 'Coco', 1, 0)
data['Kahlua'] = np.where(data['Sabor'] == 'Kahlua', 1, 0)
data['Fresa_pina'] = np.where(data['Sabor'] == 'Fresa-pina', 1, 0)
data['Sandia'] = np.where(data['Sabor'] == 'Sandia', 1, 0)
```

```python
data.head()
```

```python
from sklearn.linear_model import LinearRegression

X = np.column_stack((
    tamano,
    hielo,
    data['Naranja'],
    data['Coco'],
    data['Kahlua'],
    data['Fresa_pina'],
    data['Sandia']
))
y = data['Calificacion']

lr = LinearRegression()
lr.fit(X, y)

pred = lr.predict(X)
```

```python
import statsmodels.api as sm

# Agregar constante para el intercepto
X_OLS = sm.add_constant(X)

# Modelo OLS
modelo_OLS = sm.OLS(y, X_OLS)

# Ajustar modelo
resultado_OLS = modelo_OLS.fit()

# Ver resultados completos
print(resultado_OLS.summary())
```

```python
print('Coeficientes:')
print(resultado_OLS.params)

print('Errores estandar:')
print(resultado_OLS.bse)

print('Estadisticos t:')
print(resultado_OLS.tvalues)

print('P-values:')
print(resultado_OLS.pvalues)
```

```python
print('Intercepto:', lr.intercept_)
print('Coeficientes:', lr.coef_)
print('R2:', lr.score(X, y))
```

```python
# Errores
error = y - pred

# RSS
RSS = np.sum(error**2)

# Numero de datos
n = len(y)

# Numero de parametros
p = X.shape[1] + 1

# Varianza del error
sigma2 = RSS / (n-p)

print('RSS:', RSS)
print('Varianza del error:', sigma2)
```

```python
# Tamano
encoder_tamano = OrdinalEncoder(categories=[['chica','grande']])
tamano = encoder_tamano.fit_transform(data[['Tamano']])

# Hielo
encoder_hielo = OrdinalEncoder(categories=[['no','si']])
hielo = encoder_hielo.fit_transform(data[['Hielo']])

# Sabor
encoder_sabor = OneHotEncoder(sparse_output=False)
sabores = encoder_sabor.fit_transform(data[['Sabor']])
```

```python
X = np.column_stack((tamano, hielo, sabores))

y = data['Calificacion']
print(X)
print(y)
```

```python
lr = LinearRegression()
lr.fit(X,y)

print('Intercepto:', lr.intercept_)
print('Coeficientes:', lr.coef_)
print('R2:', lr.score(X,y))
```
