```python
X_m=np.hstack([v1, x_v, x2_v])
print(X_m)
```

```python
#Cuanod usamos la libreria de sklearn no es necesario agregar la columna de unos, ya que la libreria lo hace automaticamente. Por lo tanto, podemos usar la siguiente forma de crear la matriz de diseño:
```
